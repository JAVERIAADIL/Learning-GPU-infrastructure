```python
import numpy as np
```

```python
input = np.random.randn(2,4)
#here d_model is 4 means we have 2 words representing in 4 dimension
# num of heads must be a number from which d_model is divisible other wise we donot get back to original shape of vector after splitting them between heads and also if it is undivisble there is soem reminder it is difficult to know head dimension
d_model = 4
num_of_heads = 2 # as 4 is divisble by 2
d_k = int(d_model / num_of_heads)
```

```python
#since we have 2 heads so we have 2 wq wk and wv
W0 = np.random.randn(d_model, d_model)
# we give w0 dimension as d_model, d_model because at end we get same dimension as input i.e 2,4 so for multiply c=r and we need to  * output with w0 so w0 must be (d_model, anything) so we make it standard and use as d)model, d_model
wq1 = np.random.randn(d_model, d_k)
wk1 = np.random.randn(d_model, d_k)
wv1 = np.random.randn(d_model, d_k)
wq2 = np.random.randn(d_model, d_k)
wk2 = np.random.randn(d_model, d_k)
wv2 = np.random.randn(d_model, d_k)
```

```python
def matrixMul(input, wq1, wq2, wk1, wk2, wv1,wv2):
  Q1 = input @ wq1 #numpy @ use for matrix multiplication
  Q2 = input @ wq2
  K1 = input @ wk1
  K2 = input @ wk2
  V1 = input @ wv1
  V2 = input @ wv2

  return Q1, Q2, K1, K2, V1, V2
```

```python
Q1, Q2, K1, K2, V1, V2 = matrixMul(input, wq1, wq2, wk1, wk2, wv1, wv2)
```

```python
def attention(Q, K ,V):
  multiply = (Q @ K.T)/np.sqrt(K.ndim)
  softmax = np.exp(multiply)/np.sum(np.exp(multiply) , axis = 1, keepdims=True)
  output = softmax @ V
  return multiply, softmax, output
```

```python
multiply, softmax, head_1 = attention( Q1, K1, V1)
multiply, softmax, head_2 = attention( Q2, K2, V2)
print(f" Head_1 Dimesion :{head_1.shape} Head_2 Dimension:{ head_2.shape}")
op = np.concatenate((head_1, head_2), axis = 1)
print(f" Output Dimension : {op.shape}")
final_output = op @  W0
print(f" Final output : {final_output} , Fianl output Dimension: {final_output.shape}")

```

```python
import numpy as np
```

```python
input_1 = np.random.randn(16,4)
Wq = np.random.randn(4,8)
Wk = np.random.randn(4,8)
Wv = np.random.randn(4,8)
```

here (16, 4) means 16 values on y axis means 16 rows total and 4 values on xaxis means 4 columns total so for multiplying one row of input1 which contain 4 values and wq wk wv 4 rows means 4 values in y axis means one column have 4 values so both metric input1 row(4 values) and wq column(4 values) have same size so they can multiply

since we have to multiply unput to wq, wv, wk so input rows must be equal to their columns . wq, wk wv may have different sizes but we have to multiply later q and k also so if wq wk are different in size then k q row column donot match so we keep their size same

```python
def matrixMul(input_1, Wq, Wk, Wv):
  Q = input_1 @ Wq #numpy @ use for matrix multiplication
  K = input_1 @ Wk
  V = input_1 @ Wv

  return Q, K, V

```

```python
Q,K, V = matrixMul(input_1, Wq, Wk, Wv)
print(f"Q: {Q} , K:{K}, V:{V}")
```

```python

print(f" Q shape: {Q.shape}, K shape: {K.shape}, V shape: {V.shape} ")
```

```python
def softmax(scores):
  return np.exp(scores)/np.sum(np.exp(scores))
```

```python
def selfattention(Q,K, V):
  scores = (Q @ K.T)/np.sqrt(K.ndim)
  weights = softmax(scores)
  output = weights @ V
  return scores, weights, output
```

```python
scores, weights, output = selfattention(Q,K, V)
print(f"scores: {scores}, weights: {weights}, output: {output}")
```
